## Задача 1
$N$ хоббитов делят $K$ кусков эльфийского хлеба поровну, не делящийся нацело остаток остается в корзинке у Сэма. Напишите функцию, которая принимает на вход параметры $N$ и $K$ и возвращает два числа: $x$ - cколько кусков эльфиского хлеба достанется каждому хоббиту, и $y$ - сколько кусков остаётся в корзинке.

```python
def share_bread(N, K):
    x = K // N  # Каждый хоббит получает равную долю нацело
    y = K % N   # Остаток отправляется в корзинку к Сэму
    print(f'Каждому хоббиту достаётся {x} кусков хлеба, а в корзинку Сэму возвращается {y} кусков')
    return x, y

# если в функции всё верно, то после выполнения этой строчки, не должно выскакивать ошибок
assert share_bread(N=3, K=14) == (4, 2)
```

## Задача 2

В копях Мории хоббиты нашли стену, на которой высечены разные натуральные числа. Согласно древним сказаниям, это даты сражений. Хоббиты знают, что сражения происходили только по високосным годам. Помогите хоббитам определить, является ли год с данным числом датой великого сражения. Если это так, то верните строку "YOU SHALL PASS", иначе верните "YOU SHALL NOT PASS". Напомним, что в соответствии с хоббитским календарем, год является високосным, если его номер кратен 4, но не кратен 100, а также если он кратен 400.

```python
def leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        text_result = 'YOU SHALL PASS'
    else:
        text_result = 'YOU SHALL NOT PASS'
    return text_result

assert leap_year(5) == 'YOU SHALL NOT PASS'
```

```python
assert leap_year(2000) == 'YOU SHALL PASS'
```

```python
assert leap_year(2024) == 'YOU SHALL PASS'
```

Для могущественного магического ритуала Гендальфу необходимо быстро подсчитывать площадь своего амулета, который умеет менять размеры. Известно, что амулет имеет форму треугольника и Гендальф знает длину каждой из сторон. Напишите функцию, которая считает площадь амулета по трем сторонам.  

**Подсказка:** используйте формулу Герона
$$ S = \sqrt{p(p-a)(p-b)(p-c)} $$
$$ p = \frac{a + b + c}{2}$$


```python
import math

def amulet_area(a, b, c):
    # Находим полупериметр
    p = (a + b + c) / 2

    # Считаем площадь
    S = math.sqrt(p * (p - a) * (p - b) * (p - c))
    return S

assert amulet_area(3, 4, 5) == 6
```

```python
import numpy as np
import math

def cal_euclidean(a, b):
    a, b = (a, b)
    return math.sqrt(math.fsum((ai - bi) ** 2 for ai, bi in zip(a, b)))

def cal_manhattan(a, b):
    a, b = (a, b)
    return math.fsum(abs(ai - bi) for ai, bi in zip(a, b))

def cal_cosine(a, b):
    a, b = (a, b)
    # Скалярное произведение векторов
    dot_product = math.fsum(ai * bi for ai, bi in zip(a, b))

    # Квадраты координат для расчета длин векторов
    norm_a = sum(ai ** 2 for ai in a)
    norm_b = sum(bi ** 2 for bi in b)

    # Защита от деления на ноль, если один из векторов нулевой
    if norm_a == 0 or norm_b == 0:
        raise ValueError("Косинусное расстояние нельзя посчитать для нулевого вектора")

    # Длины векторов 
    length_a = math.sqrt(norm_a)
    length_b = math.sqrt(norm_b)
            
    # Косинусное расстояние = 1 - косинусное сходство
    distance = 1 - (dot_product / (length_a * length_b))

    return distance
```

```python
a = np.random.randint(-10, 10, size=10)
b = np.random.randint(-10, 10, size=10)
print(cal_euclidean(a, b))
print(cal_manhattan(a, b))
print(cal_cosine(a, b))
```

Создайте случайный array (`np.random.rand()`) длиной 100. Преобразуйте его так, чтобы
* Максимальный элемент(ы) был равен 1
* Минимальный элемент(ы) был равен 0
* Остальные элементы в интервале от 0 до 1 остаются прежними

```python
import numpy as np

# Создаем случайный массив из 100 элементов в диапазоне [0, 1)
my_array = np.random.rand(100)

# Находим текущие минимальное и максимальное значения
arr_min = my_array.min()
arr_max = my_array.max()

# Меняем минимальный элемент на 0 и максимальный на 1
my_array[my_array == arr_min] = 0.0
my_array[my_array == arr_max] = 1.0

print(np.max(my_array), np.min(my_array))
print(my_array)
```

Напишите функцию, которая принимает на вход матрицу (array) X и возвращает все её уникальные строки в виде новой матрицы.

```python
import numpy as np

def get_unique_rows(X):
    # Находим уникальные строки
    X_unique = np.unique(X, axis=0)
    return X_unique
```

```python
X = np.random.randint(4, 6, size=(10,3))
X_unique = get_unique_rows(X)

print(f"Исходная матрица X:\n{X}\n")
print(f"Уникальные строки матрицы X:\n{X_unique}\n")
```
