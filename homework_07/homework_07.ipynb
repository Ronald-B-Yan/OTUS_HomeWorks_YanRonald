### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
# ВАШ КОД ЗДЕСЬ
# Создание одномерного массива из 10 нулей
array_1d = np.zeros(10, dtype=int)

# Замена 5-го элемента на 1
array_1d[4] = 1
print(f'Одномерный массив:\n{array_1d}')

# Преобразование в двумерный массив размерностью 2x5
array_2d_v1 = array_1d.reshape(2, 5)
print(f'\nДвумерный массив (размер 2x5):\n{array_2d_v1}')

# Преобразование в двумерный массив размерностью 5x2
array_2d_v2 = array_1d.reshape(5, 2)
print(f'\nДвумерный массив (размер 5x2):\n{array_2d_v2}')
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
# ВАШ КОД ЗДЕСЬ
# Создание одномерного массива со значениями от 10 до 49
array_1d = np.arange(10, 50)
print(f'Одномерный массив:\n{array_1d}')

# Переворот массива (первый элемент становится последним)
reversed_array = array_1d[::-1]
print(f'\nРазвёрнутый массив:\n{reversed_array}')

# Выбор всех чётных элементов
even_elements = reversed_array[reversed_array % 2 == 0]
print(f'\nЧётные элементы из развернутого массива:\n{even_elements}')
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
# ВАШ КОД ЗДЕСЬ
# Создание трехмерного массива размером 4x3x2 со случайными числами
array_3d = np.random.rand(4, 3, 2)
print(f'\nИсходный массив 4x3x2:\n{array_3d}')


# Поиск минимального и максимального значения в массиве
min_value = array_3d.min()
max_value = array_3d.max()

print(f'\nМинимальное значение в массиве: {min_value:.3f}')
print(f'Максимальное значение в массиве: {max_value:.3f}')
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
# Создание двух двумерных массивов 6x4 и 4x3
# Матрица A размером 6x4
matrix_a = np.random.rand(6, 4)

# Матрица B размером 4x3
matrix_b = np.random.rand(4, 3)

# Матричное умножение
matrix_result = matrix_a @ matrix_b

print(f'Матрица A (6x4):\n{matrix_a}')
print(f'\nМатрица B (4x3):\n{matrix_b}')
print(f'\nМатрица - результат умножения (6x3):\n{matrix_result}')
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
# ВАШ КОД ЗДЕСЬ
# Создание двумерного массива 7х7
matrix_с = np.random.rand(7, 7)
print(f'Матрица С (7x7):\n{matrix_с}')

# Поиск среднего и стандартного отклонения массива
mean_value = matrix_с.mean()
std_value = matrix_с.std()
print(f'\nСреднее значение в массиве: {mean_value:.3f}')
print(f'Стандартное отклонение значений в массиве: {std_value:.3f}')


# Нормализация массива (Z-score нормализация / стандартизация) по формуле: (x - mean) / std
matrix_с_norm = (matrix_с - mean_value) / std_value
print('\nНормализованный массив:\n', matrix_с_norm)

# Проверка среднего и стандартного отклонения нормализованного массива
mean_value_norm = matrix_с_norm.mean()
std_value_norm = matrix_с_norm.std()
print(f'\nСреднее значение в массиве: {mean_value_norm:.3f}')
print(f'Стандартное отклонение значений в массиве: {std_value_norm:.3f}')
```

### Найдите количество курящих людей

```python
# ВАШ КОД ЗДЕСЬ
smokers_count = df['smoker'].value_counts().get('Yes')
print(f'Количество курящих людей: {smokers_count}')
```

### Узнайте какой средний 'total_bill' в зависимости от 'day'

```python
# ВАШ КОД ЗДЕСЬ
mean_bill_by_day = df.groupby('day')['total_bill'].mean()
print(f'Средний счёт в зависимости от дня недели {mean_bill_by_day.round(1)}')
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
# ВАШ КОД ЗДЕСЬ
# Находим медианное значение счёта
median_bill = df['total_bill'].median()
print(f'Медианное значение счёта: {median_bill.round(1)}\n')

# Отбираем строчки больше медианы
filtered_df_by_total_bill = df[df['total_bill'] > median_bill]

# Считаем средние чаевые в зависимости от пола
mean_tip_by_sex = filtered_df_by_total_bill.groupby('sex')['tip'].mean()
print(f'Средние чаевые в зависимости от пола {mean_tip_by_sex.round(2)}')
```

### Преобразуйте признак 'smoker' в бинарный (0-No, 1-Yes)

```python
# ВАШ КОД ЗДЕСЬ
df['smoker'] = df['smoker'].replace({'Yes': 1, 'No': 0})
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
# ВАШ КОД ЗДЕСЬ
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Гистограмма для Lunch (Обед)
sns.histplot(
    data=df[df['time'] == 'Lunch'],
    x='tip',
    kde=True,
    bins=15,
    color='orange',
    ax=axes[0],
)
axes[0].set_title('Обед (Lunch)', fontsize=10)
axes[0].set_xlabel('Чаевые ($)')
axes[0].set_ylabel('Количество')

# Гистограмма для Dinner (Ужин)
sns.histplot(
    data=df[df['time'] == 'Dinner'],
    x='tip',
    kde=True,
    bins=15,
    color='purple',
    ax=axes[1],
)
axes[1].set_title('Ужин (Dinner)', fontsize=10)
axes[1].set_xlabel('Чаевые ($)')
axes[1].set_ylabel('Количество')

plt.suptitle('Сравнение распределения чаевых: Обед vs Ужин', fontsize=12)
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# График для Мужчин
sns.scatterplot(
    data=df[df['sex'] == 'Male'], 
    x='total_bill', 
    y='tip', 
    hue='smoker', 
    palette='Set1',
    ax=axes[0]
)
axes[0].set_title('Мужчины (Male)', fontsize=10)
axes[0].set_xlabel('Total Bill')
axes[0].set_ylabel('Tip')

# График для Женщин
sns.scatterplot(
    data=df[df['sex'] == 'Female'], 
    x='total_bill', 
    y='tip', 
    hue='smoker', 
    palette='Set1',
    ax=axes[1]
)
axes[1].set_title('Женщины (Female)', fontsize=10)
axes[1].set_xlabel('Total Bill')
axes[1].set_ylabel('Tip')

plt.suptitle("Взаимосвязь total_bill и tip в зависимости от smoker", fontsize=12)
```

## Сделайте выводы по анализу датасета и построенным графикам. По желанию можете продолжить анализ данных и также отразить это в выводах.

```python
# Создадим новй признак - процент чаевых
df['tip_percent'] = (df['tip'] / df['total_bill']) * 100
print(f'Средний процент чаевых:, {df['tip_percent'].mean():.1f}%')
```

```python
# ВАШ КОД ЗДЕСЬ
# Посмотрим, как зависит размер чаевых от размера компании за столом

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Строим график зависимости размера чаевых от размера компании
sns.lineplot(
    data=df, 
    x='size', 
    y='tip', 
    marker='o', 
    color='purple',
    ax=axes[0],
    )

axes[0].set_title('Размер чаевых', fontsize=10)
axes[0].set_xlabel('Размер компании')
axes[0].set_ylabel('Tip')


# Строим график зависимости процентов чаевых от размера компании
sns.lineplot(
    data=df, 
    x='size', 
    y='tip_percent', 
    marker='o', 
    color='purple',
    ax=axes[1],
    )

axes[1].set_title('% чаевых от суммы заказа', fontsize=10)
axes[1].set_xlabel('Размер компании')
axes[1].set_ylabel('Tip')

plt.suptitle('Взаимосвязь чаевых от размера компании', fontsize=12)
```
