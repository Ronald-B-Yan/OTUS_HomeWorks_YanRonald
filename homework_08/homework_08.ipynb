## Последовательность вычислений (аналитическое решение)

### Шаг 1: Транспонирование матриц $A$ и $B$
Матрица $A^\top$ получается заменой строк на столбцы:
$$A^\top = \begin{pmatrix} 2 & 3 & -1 \\ -4 & 5 & 0 \end{pmatrix}$$

Матрица $B^\top$:
$$B^\top = \begin{pmatrix} 1 & -3 & 5 \\ 2 & -4 & 2 \\ 7 & 0 & 1 \end{pmatrix}$$

---

### Шаг 2: Вычисление произведения $A^\top C$
Умножаем матрицу $A^\top$ размера $(2 \times 3)$ на матрицу $C$ размера $(3 \times 3)$:
$$A^\top C = \begin{pmatrix} 2 & 3 & -1 \\ -4 & 5 & 0 \end{pmatrix} \begin{pmatrix} 6 & -3 & 9 \\ 4 & -5 & 2 \\ 8 & 1 & 5 \end{pmatrix}$$

$$A^\top C = \begin{pmatrix} 
2\cdot6 + 3\cdot4 + (-1)\cdot8 & 2\cdot(-3) + 3\cdot(-5) + (-1)\cdot1 & 2\cdot9 + 3\cdot2 + (-1)\cdot5 \\ 
-4\cdot6 + 5\cdot4 + 0\cdot8 & -4\cdot(-3) + 5\cdot(-5) + 0\cdot1 & -4\cdot9 + 5\cdot2 + 0\cdot5 
\end{pmatrix}$$

$$A^\top C = \begin{pmatrix} 
16 & -22 & 19 \\ -4 & -13 & -26 \end{pmatrix}$$

---

### Шаг 3: Вычисление произведения $A^\top B^\top$
Умножаем матрицу $A^\top$ размера $(2 \times 3)$ на матрицу $B^\top$ размера $(3 \times 3)$:
$$A^\top B^\top = \begin{pmatrix} 2 & 3 & -1 \\ -4 & 5 & 0 \end{pmatrix} \begin{pmatrix} 1 & -3 & 5 \\ 2 & -4 & 2 \\ 7 & 0 & 1 \end{pmatrix}$$

$$A^\top B^\top = \begin{pmatrix} 
2\cdot1 + 3\cdot2 + (-1)\cdot7 & 2\cdot(-3) + 3\cdot(-4) + (-1)\cdot0 & 2\cdot5 + 3\cdot2 + (-1)\cdot1 \\ 
-4\cdot1 + 5\cdot2 + 0\cdot7 & -4\cdot(-3) + 5\cdot(-4) + 0\cdot0 & -4\cdot5 + 5\cdot2 + 0\cdot1 
\end{pmatrix}$$

$$A^\top B^\top = \begin{pmatrix} 
1 & -18 & 15 \\ 6 & -8 & -10 \end{pmatrix}$$

---

### Шаг 4: Умножение матрицы $A^\top B^\top$ на скаляр 2
$$2A^\top B^\top = 2 \cdot \begin{pmatrix} 1 & -18 & 15 \\ 6 & -8 & -10 \end{pmatrix} = \begin{pmatrix} 2 & -36 & 30 \\ 12 & -16 & -20 \end{pmatrix}$$

---

### Шаг 5: Нахождение итоговой матрицы $D$
Вычитаем из результата Шага 2 результат Шага 4:
$$D = A^\top C - 2A^\top B^\top = \begin{pmatrix} 16 & -22 & 19 \\ -4 & -13 & -26 \end{pmatrix} - \begin{pmatrix} 2 & -36 & 30 \\ 12 & -16 & -20 \end{pmatrix}$$

$$D = \begin{pmatrix} 
16 - 2 & -22 - (-36) & 19 - 30 \\ 
-4 - 12 & -13 - (-16) & -26 - (-20) 
\end{pmatrix} = \begin{pmatrix} 14 & 14 & -11 \\ -16 & 3 & -6 \end{pmatrix}$$


```python
import numpy as np

# Инициализация исходных матриц
A = np.array([
    [2, -4],
    [3, 5],
    [-1, 0]
])

B = np.array([
    [1, 2, 7],
    [-3, -4, 0],
    [5, 2, 1]
])

C = np.array([
    [6, -3, 9],
    [4, -5, 2],
    [8, 1, 5]
])

# Вычисление транспонированных матриц
A_T = A.T
B_T = B.T

# Перемножение матриц
AT_C = A_T @ C
AT_BT = A_T @ B_T

# Итоговая матрица D
D = AT_C - 2 * AT_BT

print(f'Результат вычисления матрицы D:\n {D}')
```

### Задание 3.

Дано выражение:

$$
3 \cdot \begin{pmatrix}
x & 2 & 3 \\
-1 & y & 4
\end{pmatrix}
+ 2 \cdot \begin{pmatrix}
1 & 2 & -5 \\
2 & -6 & z
\end{pmatrix}
= \begin{pmatrix}
8 & v & -1 \\
1 & 6 & 4
\end{pmatrix}.
$$

Найдите значения $(x, y, z)$ и $v$, при которых выражение верно.

Чтобы найти значения переменных $x, y, z$ и $v$ составим систему линейных уравнений.

1. **Для переменной x (1, 1):**
   $$ 3x + 2 \cdot 1 = 8 \implies 3x = 6 \implies x = 2 $$

2. **Для переменной $v$ (1, 2):**
   $$ 3 \cdot 2 + 2 \cdot 2 = v \implies v = 10 $$

3. **Для переменной y (2, 2):**
   $$ 3y + 2 \cdot (-6) = 6 \implies 3y = 18 \implies y = 6 $$

4. **Для переменной z (2, 3):**
   $$ 3 \cdot 4 + 2z = 4 \implies 2z = -8 \implies z = -4 $$

**Ответ:**  
$x = 2$  
$y = 6$  
$z = -4$  
$v = 10$

```python
# Задаем исходные матрицы, вместо неизвестных ставим None, чтобы сохранить структуру 2x3

A = [
    [None, 2, 3],
    [-1, None, 4]
]

B = [
    [1, 2, -5],
    [2, -6, None]
]

C = [
    [8, None, -1],
    [1, 6, 4]
]

# Находим переменные по их координатам [индекс_строки][индекс_столбца]

# Находим x (строка 0, столбец 0)
x = (C[0][0] - 2 * B[0][0]) / 3

# Находим v (строка 0, столбец 1)
v = 3 * A[0][1] + 2 * B[0][1]

# Находим y (строка 1, столбец 1)
y = (C[1][1] - 2 * B[1][1]) / 3

# Находим z (строка 1, столбец 2)
z = (C[1][2] - 3 * A[1][2]) / 2

# Приводим к целым числам (int)
x, y, z, v = int(x), int(y), int(z), int(v)

# Выводим результат
print('Найденные значения переменных:')
print(f'x = {x}')
print(f'y = {y}')
print(f'z = {z}')
print(f'v = {v}')
```

### Задание 6.



* Запишите матрично-векторное уравнение $Ax = b$ в виде системы линейных уравнений.

* Решите его любым способом, найдя **вектор** $x$. Решение опишите.



$$
A =
 \begin{pmatrix}
  1 & 2 & 4 \\
  2 & 1 & -1 \\
  1 & -1 & 1 \\
 \end{pmatrix}
 $$

 $$
 b =
  \begin{pmatrix}

  -9 \\
  0 \\
 \end{pmatrix}
 $$

### Решение матрично-векторного уравнения $Ax = b$

Пусть вектор неизвестных $x = \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}$.

## Запишем систему линейных уравнений
Перемножив матрицу $A$ на вектор $x$ и приравняв к $b$, получаем систему:

$$
\begin{cases}
1\cdot x_1 + 2\cdot x_2 + 4\cdot x_3 = 3 \\
2\cdot x_1 + 1\cdot x_2 - 1\cdot x_3 = -9 \\
1\cdot x_1 - 1\cdot x_2 + 1\cdot x_3 = 0
\end{cases}
$$

### Аналитическое решение
Из третьего уравнения выразим $x_1$:
$$x_1 = x_2 - x_3$$

Подставим полученное выражение в первое и второе уравнения:
1) $(x_2 - x_3) + 2x_2 + 4x_3 = 3 \implies 3x_2 + 3x_3 = 3 \implies x_2 + x_3 = 1$
2) $2(x_2 - x_3) + x_2 - x_3 = -9 \implies 3x_2 - 3x_3 = -9 \implies x_2 - x_3 = -3$

Получили систему из двух уравнений:
$$
\begin{cases}
x_2 + x_3 = 1 \\
x_2 - x_3 = -3
\end{cases}
$$

Сложим эти два уравнения:
$$2x_2 = -2 \implies x_2 = -1$$

Подставим $x_2 = -1$ в уравнение $x_2 + x_3 = 1$:
$$-1 + x_3 = 1 \implies x_3 = 2$$

Теперь найдем $x_1$:
$$x_1 = x_2 - x_3 = -1 - 2 = -3$$

**Ответ:** $x = \begin{pmatrix} -3 \\ -1 \\ 2 \end{pmatrix}$


```python
import numpy as np

# Определяем матрицу A и вектор b
A = np.array(
    [[1, 2, 4],
     [2, 1, -1],
     [1, -1, 1]])

b = np.array([3, -9, 0])

# Решаем систему с помощью встроенной функции библиотеки NumPy
x = np.linalg.solve(A, b)

print('Программное решение системы:')
for i, val in enumerate(x, 1):
    print(f'x{i} = {val:.0f}')

# Проверка решения (Ax должно быть равно b)
print(f'\nПроверка (A * x): {A @ x}' )
```

### Задача 6-2: Функция распределения (CDF)

Функция распределения $F(x)$ определяется как интеграл от функции плотности $p(t)$ от $-\infty$ до $x$:
$$F(x) = \int_{-\infty}^{x} p(t) \, dt$$

Разделим вычисление на три интервала:
1. При $x < 0$: плотность распределения равна $0$, поэтому $F(x) = 0$.
2. При $0 \le x \le 1$:
   $$F(x) = \int_{0}^{x} 2t \, dt = \left[ t^2 \right]_0^x = x^2$$
3. При $x > 1$: случайная величина принимает значение $0$, поэтому $F(x) = 0$.

**Итоговая функция распределения (CDF):**

$$F(x) = \begin{cases} 
0, & x < 0 \\ 
x^2, & 0 \le x \le 1 \\ 
0, & x > 1 
\end{cases}$$
